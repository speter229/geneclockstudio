```python
import pandas as pd

#read in the beta values for each dataset
GSE37754_df = pd.read_csv('GSE37754_betas.csv', index_col=0)
GSE49149_df = pd.read_csv('GSE49149_betas.csv', index_col=0)
GSE52068_df = pd.read_csv('GSE52068_betas.csv', index_col=0)
GSE59157_df = pd.read_csv('GSE59157_betas.csv', index_col=0)
GSE77955_df = pd.read_csv('GSE77955_betas.csv', index_col=0)
```

```python
#cg intersection calculation
set1=set(GSE37754_df.index.tolist())
set2=set(GSE49149_df.index.tolist())
set3=set(GSE52068_df.index.tolist())
set4=set(GSE59157_df.index.tolist())
set5=set(GSE77955_df.index.tolist())

cg_intersection = list(set1.intersection(set2, set3, set4, set5))
# Display the result
print("Intersection length of the sets:", len(cg_intersection)) #402496   #without GSE49149_df it would be 485512 (but it contain 196 important samples)
```

```python
#KEEPING ONLY THE COMMON CPG SITES
GSE37754_df = GSE37754_df.loc[cg_intersection]
GSE49149_df = GSE49149_df.loc[cg_intersection]
GSE52068_df = GSE52068_df.loc[cg_intersection]
GSE59157_df = GSE59157_df.loc[cg_intersection]
GSE77955_df = GSE77955_df.loc[cg_intersection]
```

```python
#NOW ORDERING THE ROWS IN THE SAME ORDER EVERYWHERE, FOR EXAMPLE INCREASING ORDER
GSE37754_df=GSE37754_df.sort_index()
GSE49149_df=GSE49149_df.sort_index()
GSE52068_df=GSE52068_df.sort_index()
GSE59157_df=GSE59157_df.sort_index()
GSE77955_df=GSE77955_df.sort_index()
```

```python
#CREATE A DATAFRAME FROM THEM SINCE NOW ALL OF THEM HAS THE SAME CG SITES IN THE SAME ORDER
# List of DataFrames to concatenate
dfs = [GSE37754_df, GSE49149_df, GSE52068_df, GSE59157_df, GSE77955_df]
# Concatenate DataFrames along columns (axis=1)
betas_combined = pd.concat(dfs, axis=1)
```
